1. Prior Probabilities

started by calculating the prior probabilities based on the dataset:

P(Yes) = 9 / 14 ≈ 0.6429

P(No) = 5 / 14 ≈ 0.3571

These represent the probabilities of playing or not playing without considering any weather features.

2. Conditional Probabilities

P(Sunny | No) = 3 / 5

P(Overcast | No) = 0 / 5

P(Rainy | No) = 2 / 5

Temperature

Humidity

Windy
