# Naive Bayes

***You are building a simple spam detector.***
***Each email is described by three features:***

***Offer: Yes or No (mentions a special offer)***

***Length: Short or Long***

***Link: Yes or No (contains a clickable link)***

***You collected the following training data (10 emails):***

#  Offer  Length  Link   Class
1  Yes    Short   Yes    Spam
2  Yes    Short   Yes    Spam
3  No     Long    No     NotSpam
4  Yes    Long    Yes    Spam
5  No     Short   No     NotSpam
6  No     Long    No     NotSpam
7  Yes    Short   No     Spam
8  No     Short   Yes    NotSpam
9  No     Long    Yes    NotSpam
10 Yes    Long    No     Spam


***A new email arrives:***

Offer = Yes
Length = Long
Link = Yes
Classify this email as Spam or NotSpam using Naive Bayes with Laplace smoothing (add-1)

There are 5 Spam emails and 5 NotSpam emails.

P(Spam)    = 5/10 = 0.5
P(NotSpam) = 5/10 = 0.5


Each feature has 2 possible values, so V = 2.
Formula:    P(value | class) = (count + 1) / (N_class + 2)
where N_class = number of training emails in that class.

Counts for Spam (N_class = 5):
Offer=Yes   count=5  → (5+1)/(5+2) = 6/7 ≈ 0.857
Length=Long count=2  → (2+1)/(5+2) = 3/7 ≈ 0.429
Link=Yes    count=3  → (3+1)/(5+2) = 4/7 ≈ 0.571


Counts for NotSpam (N_class = 5):
Offer=Yes   count=0  → (0+1)/(5+2) = 1/7 ≈ 0.143
Length=Long count=3  → (3+1)/(5+2) = 4/7 ≈ 0.571
Link=Yes    count=2  → (2+1)/(5+2) = 3/7 ≈ 0.429


Multiply the prior by all conditional probabilities.

For Spam:

        score(Spam) =
0.5 × (6/7) × (3/7) × (4/7)
= 0.5 × 72/343
= 36/343
≈ 0.105





For NotSpam:

         score(NotSpam) =
0.5 × (1/7) × (4/7) × (3/7)
= 0.5 × 12/343
= 6/343
≈ 0.017


Normalize the scores:
                P(Spam | x)    = 0.105 / (0.105 + 0.017)
               = 36/42
               ≈ 0.857

P(NotSpam | x) = 0.017 / (0.105 + 0.017)
               = 6/42
               ≈ 0.143


Since P(Spam | x) is much higher:


Final decision: The email is Spam.
