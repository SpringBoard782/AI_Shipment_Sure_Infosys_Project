```python
'''1.Use map to create a new column: Shipment_Code (Flight→1, Ship→2, Road→3)'''
shipment_map = {'Flight': 1, 'Ship': 2, 'Road': 3}
df['Shipment_Code'] = df['Mode_of_Shipment'].map(shipment_map)
df
```

```python
'''2.Group by Gender and calculate the percentage of on-time deliveries'''
(df.groupby('Gender')['Reached.on.Time_Y.N'].mean() * 100).round(2)
```

```python
'''3.Filter rows where Product_importance is 'high' and Cost_of_the_Product > 250'''
df.loc[(df['Product_importance'] == 'high') & (df['Cost_of_the_Product'] > 250)]
```

```python
'''4.Use apply to create a new column: Call_Frequency (Frequent if calls > 3, else Not frequent)'''
df['Call_Frequency'] = df['Customer_care_calls'].apply(lambda x: 'Frequent' if x > 3 else 'Not frequent')
df
```

```python
'''5. Filter rows where Weight_in_gms is an outlier '''
Q1 = df['Weight_in_gms'].quantile(0.25)
Q3 = df['Weight_in_gms'].quantile(0.75)
IQR = Q3 - Q1
df.loc[(df['Weight_in_gms'] < (Q1 - 1.5 * IQR)) | (df['Weight_in_gms'] > (Q3 + 1.5 * IQR))]
```

```python
'''6.Map Customer_rating to a descriptive string (e.g., 1 → 'Poor', 5 → 'Excellent')'''
rating_map = {1: 'Poor', 2: 'Fair', 3: 'Good', 4: 'Very Good', 5: 'Excellent'}
df['Rating_Description'] = df['Customer_rating'].map(rating_map)
df
```

```python
'''7.Group by Product_importance and Gender, then count occurrences'''
df.groupby(['Product_importance', 'Gender']).size().unstack(fill_value=0)
```

```python
'''8.Filter rows where Discount_offered is in the top 5%'''
top_5_percent = df['Discount_offered'].quantile(0.95)
df.loc[df['Discount_offered'] >= top_5_percent]
```

```python
'''9.Use iloc to select every 10th row and all columns'''
df.iloc[::10, :]
```

```python
'''10.Group by Mode_of_Shipment and Product_importance, then calculate mean cost'''
df.groupby(['Mode_of_Shipment', 'Product_importance'])['Cost_of_the_Product'].mean().unstack()
```
