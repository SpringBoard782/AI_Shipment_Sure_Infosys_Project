```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Load the dataset
try:
    df = pd.read_csv('C:\\Users\\pavan\\downloads\\Train.csv')
except FileNotFoundError:
    print("Error: The file 'Train.csv' was not found. Please provide the correct file path.")
    exit()

# -----------------------------------------------------------------------------
# 1. Bar Plot: On-Time Delivery Rate by Mode of Shipment
# FIX: The 'palette' warning is fixed by assigning 'hue' and setting 'legend=False'.
# -----------------------------------------------------------------------------
delivery_rate = df.groupby('Mode_of_Shipment')['Reached.on.Time_Y.N'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(
    data=delivery_rate,
    x='Mode_of_Shipment',
    y='Reached.on.Time_Y.N',
    hue='Mode_of_Shipment', # Assign the x-variable to hue
    palette='muted',
    legend=False # Hide the redundant legend
)
plt.title('On-Time Delivery Rate by Mode of Shipment', fontsize=16)
plt.xlabel('Mode of Shipment', fontsize=12)
plt.ylabel('On-Time Delivery Rate', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, 1)
plt.tight_layout()
plt.savefig('on_time_delivery_rate.png')
plt.show()
```

```python
# -----------------------------------------------------------------------------
# 2. Violin Plot: Weight Distribution by Gender and On-Time Delivery
# Violin plots show the full distribution shape of a numerical variable across
# categories, including the median and quartiles (like a box plot).
# -----------------------------------------------------------------------------
plt.figure(figsize=(6, 6))
sns.violinplot(
    data=df,
    x='Gender',
    y='Weight_in_gms',
    hue='Reached.on.Time_Y.N',
    split=True,
    palette={1: 'green', 0: 'red'}
)
plt.title('Weight Distribution by Gender and On-Time Delivery', fontsize=16)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Weight (grams)', fontsize=12)
plt.legend(title='Reached on Time', labels=['No', 'Yes'])
plt.tight_layout()
plt.savefig('weight_violin_plot.png')
plt.show()
```

```python
# -----------------------------------------------------------------------------
# 3. Joint Plot: Cost of Product vs. Weight
# A joint plot shows the relationship between two variables and their
# individual histograms on the margins. This is great for bivariate analysis.
# -----------------------------------------------------------------------------
sns.set_style("white") # Set a clean background for the joint plot
joint_plot = sns.jointplot(
    data=df,
    x='Cost_of_the_Product',
    y='Weight_in_gms',
    kind='reg', # 'reg' adds a regression line and confidence interval
    color='purple',
    height=5
)
joint_plot.set_axis_labels('Cost of the Product ($)', 'Weight (grams)', fontsize=7)
plt.suptitle('Cost of Product vs. Weight in Gms', y=1.02, fontsize=6)
plt.tight_layout()
plt.savefig('cost_vs_weight_jointplot.png')
plt.show()
```

```python

# 1.Add 'Total_Cost_with_Discount' column
# This new column calculates the final cost of the product after the discount is applied.
# The formula is: Original Cost * (1 - Discount Percentage)

df['Total_Cost_with_Discount'] = df['Cost_of_the_Product'] * (1 - df['Discount_offered'] / 100)
print(df)
```

```python

# 2. Add 'Is_Heavy_Package' column
# This is a boolean (True/False) column that checks if the weight is over 2000 grams.
# This can be useful for logistics analysis.

df['Is_Heavy_Package'] = df['Weight_in_gms'] > 2000
print(df)
```

```python

# 3. Add 'Overall_Service_Rating' column
# This new categorical column combines two existing metrics: 'Customer_rating' and
# 'Reached.on.Time_Y.N' to create a single, simplified rating.

def get_service_rating(row):
    """
    Function to determine the overall service rating based on customer rating
    and on-time delivery.
    """
    on_time = row['Reached.on.Time_Y.N']
    rating = row['Customer_rating']

    if on_time == 1 and rating >= 4:
        return 'Excellent'
    elif on_time == 1 and rating < 4:
        return 'Good'
    elif on_time == 0 and rating >= 4:
        return 'Poor'
    else: # on_time == 0 and rating < 4
        return 'Bad'

df['Overall_Service_Rating'] = df.apply(get_service_rating, axis=1)
print(df)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, f1_score, recall_score

# Load the dataset
df = pd.read_csv('C:\\Users\\pavan\\downloads\\Train.csv')

# Drop the 'ID' column as it is not a feature for the model
df = df.drop('ID', axis=1)

# Define the target variable and features
target = 'Reached.on.Time_Y.N'
features = df.drop(target, axis=1)
y = df[target]

# Identify categorical and numerical features
categorical_features = ['Warehouse_block', 'Mode_of_Shipment', 'Product_importance', 'Gender']
numerical_features = features.drop(columns=categorical_features).columns

# Create preprocessing pipelines for numerical and categorical features
numeric_transformer = StandardScaler()
categorical_transformer = OneHotEncoder(handle_unknown='ignore')

# Create a preprocessor using ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ])

# Create the logistic regression model pipeline
model_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                                 ('classifier', LogisticRegression(random_state=42))])

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(features, y, test_size=0.2, random_state=42)

# Train the model
model_pipeline.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model_pipeline.predict(X_test)

# Calculate and print the evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)

print(f'Model accuracy: {accuracy:.2f}')
print(f'Model F1-score: {f1:.2f}')
print(f'Model recall: {recall:.2f}')
```
