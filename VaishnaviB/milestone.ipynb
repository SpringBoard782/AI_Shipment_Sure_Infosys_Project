```python
#conditional probablity
# P(Reached on time | Rating >= 4)
p_B = len(B) / len(df)
p_A_and_B = len(intersection) / len(df)
p_A_given_B = p_A_and_B / p_B

print("P(Reached on time | Rating >= 4):", p_A_given_B)
```

```python
# Example for Bayes: P(A|B) = P(B|A) * P(A) / P(B)

p_A = len(A) / len(df)  # P(A)
p_B_given_A = len(intersection) / len(A)  # P(B|A)

bayes_p = (p_B_given_A * p_A) / p_B
print("Bayes Theorem P(A|B):", bayes_p)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import SelectKBest, chi2, RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression

# 1️⃣ Load your dataset
df = pd.read_csv("C:/Users/DELL/Downloads/archive/Train.csv")

# 2️⃣ Encode categorical variables
df_encoded = df.copy()
le = LabelEncoder()
for col in df_encoded.select_dtypes(include='object').columns:
    df_encoded[col] = le.fit_transform(df_encoded[col])

# Separate features (X) and target (y)
X = df_encoded.drop(columns=['ID', 'Reached.on.Time_Y.N'])
y = df_encoded['Reached.on.Time_Y.N']

# ================================
# METHOD 1: Correlation Analysis
# ================================
correlation_matrix = X.corrwith(y).abs().sort_values(ascending=False)

# ================================
# METHOD 2: Chi-Square Test
# ================================
chi2_selector = SelectKBest(score_func=chi2, k='all')
chi2_selector.fit(X, y)
chi2_scores = pd.Series(chi2_selector.scores_, index=X.columns).sort_values(ascending=False)

# ================================
# METHOD 3: Random Forest Importance
# ================================
rf = RandomForestClassifier(random_state=42)
rf.fit(X, y)
rf_importances = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)

# ================================
# METHOD 4: Recursive Feature Elimination (RFE)
# ================================
model = LogisticRegression(max_iter=1000)
rfe = RFE(model, n_features_to_select=5)  # choose top 5
rfe.fit(X, y)
rfe_selected = pd.Series(rfe.support_, index=X.columns)

# ================================
# Combine Results
# ================================
feature_selection_results = pd.DataFrame({
    'Correlation_with_Target': correlation_matrix,
    'Chi2_Score': chi2_scores,
    'RF_Importance': rf_importances,
    'RFE_Selected': rfe_selected
}).sort_values(by='RF_Importance', ascending=False)

# Show results
print(feature_selection_results)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import chi2, mutual_info_classif, SelectKBest, RFE
from sklearn.linear_model import LogisticRegression, LassoCV
from sklearn.ensemble import RandomForestClassifier

# 1️⃣ Load dataset
df = pd.read_csv("C:/Users/DELL/Downloads/archive/Train.csv")

# 2️⃣ Encode categorical variables; 
label_encoders = {}
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le
#Loops through all categorical columns (dtype='object').Converts them into numeric codes using LabelEncoder.Stores each encoder in label_encoders dictionary (useful if you want to decode later).


# 3️⃣ Separate features & target
target_col = 'Reached.on.Time_Y.N'  
X = df.drop(target_col, axis=1)
y = df[target_col]
#X → all independent variables.y → the dependent variable (what you’re predicting).


# 4️⃣ Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
#Standardizes data → each feature will have mean=0 and standard deviation=1.This helps methods like Logistic Regression & Lasso work better.
# Dictionary to store results
results = {}
#We’ll store the scores from each feature selection method here.



# 5️⃣ Method 1: Correlation Analysis (fixed)
corr_scores = pd.Series(
    [np.corrcoef(X_scaled[:, i], y)[0, 1] for i in range(X_scaled.shape[1])],
    index=X.columns
).abs()
results['Correlation'] = corr_scores
#Loops through each feature and calculates Pearson correlation with target..abs() → takes absolute value because we only care about strength, not sign.Stores results in results['Correlation'].


# 6️⃣ Method 2: Chi-Square Test
chi2_selector = SelectKBest(score_func=chi2, k='all')
chi2_selector.fit(X, y)
results['Chi2'] = pd.Series(chi2_selector.scores_, index=X.columns)
#Runs Chi-Square test between each feature & target.
#Works best when features are non-negative (raw values, not scaled).
#Higher score = stronger relationship.


# 7️⃣ Method 3: Mutual Information
mi_selector = SelectKBest(score_func=mutual_info_classif, k='all')
mi_selector.fit(X, y)
results['Mutual_Info'] = pd.Series(mi_selector.scores_, index=X.columns)
#Calculates Mutual Information (how much knowing this feature reduces uncertainty in the target).
#Captures both linear & non-linear relationships.


# 8️⃣ Method 4: Recursive Feature Elimination (RFE)
log_reg = LogisticRegression(max_iter=5000)
rfe_selector = RFE(log_reg, n_features_to_select=5)
rfe_selector.fit(X_scaled, y)
results['RFE_Rank'] = pd.Series(rfe_selector.ranking_, index=X.columns)
#Uses Logistic Regression as a model.Removes least important features step by step until only 5 remain.Rank = 1 → selected feature, higher numbers = less important.


# 9️⃣ Method 5: Random Forest Importance
rf = RandomForestClassifier(random_state=42)
rf.fit(X, y)
results['RF_Importance'] = pd.Series(rf.feature_importances_, index=X.columns)
#rains a Random Forest on your dataset.Measures how much each feature reduces prediction error (Gini importance).

# 🔟 Method 6: Lasso Regression
lasso = LassoCV(cv=5, random_state=42)
lasso.fit(X_scaled, y)
results['Lasso_Coeff'] = pd.Series(np.abs(lasso.coef_), index=X.columns)
#Runs Lasso Regression with cross-validation.Lasso penalizes unnecessary features → pushes their coefficients to 0.Higher absolute coefficient = more important.

# Combine results
feature_scores = pd.DataFrame(results)
#Creates a DataFrame with all methods' scores.Sorts it by Random Forest importance for easy viewing.
# Sort by Random Forest Importance
feature_scores_sorted = feature_scores.sort_values(by='RF_Importance', ascending=False)

print("\n=== Feature Selection Summary ===\n")
print(feature_scores_sorted)

# Save results directly to Downloads folder
save_path = r"C:\Users\DELL\Downloads\feature_selection_results.csv"
feature_scores_sorted.to_csv(save_path)
print(f"\n✅ Results saved to: {save_path}")
```

```python
#1. Filter Method (Statistical Tests).Uses correlation, Chi-square, ANOVA, etc.Selects features based on statistical measures, independent of ML model.

import pandas as pd
import numpy as np
from sklearn.feature_selection import chi2, SelectKBest, f_classif

# Load dataset
df = pd.read_csv("C:/Users/DELL/Downloads/archive/Train.csv")
df = df.drop("ID", axis=1)

# Encode categorical features
df['Product_importance'] = df['Product_importance'].map({'low':1, 'medium':2, 'high':3})
df['Gender'] = df['Gender'].map({'M':0, 'F':1})
df = pd.get_dummies(df, columns=['Warehouse_block','Mode_of_Shipment'], drop_first=True)

X = df.drop("Reached.on.Time_Y.N", axis=1)
y = df["Reached.on.Time_Y.N"]

# --- Chi-Square (for categorical/discrete features)
chi_selector = SelectKBest(score_func=chi2, k=5)
chi_selector.fit(X, y)
chi_features = X.columns[chi_selector.get_support()]
print("\n Filter Method (Chi-square) Selected Features:\n", list(chi_features))

# --- ANOVA F-test (for numerical features vs classification target)
anova_selector = SelectKBest(score_func=f_classif, k=5)
anova_selector.fit(X, y)
anova_features = X.columns[anova_selector.get_support()]
print("\nFilter Method (ANOVA F-test) Selected Features:\n", list(anova_features))
```

```python
#2. Embedded Method (Model-Based).Uses algorithms that automatically select features while training (e.g., Lasso, Tree-based models).

from sklearn.linear_model import LassoCV
from sklearn.ensemble import RandomForestClassifier

# --- Lasso Regression (L1 Regularization)
lasso = LassoCV(cv=5).fit(X, y)
lasso_coef = pd.Series(lasso.coef_, index=X.columns)
selected_lasso = list(lasso_coef[lasso_coef != 0].index)
print("\nEmbedded Method (Lasso) Selected Features:\n", selected_lasso)

# --- Random Forest Feature Importance
rf = RandomForestClassifier(random_state=42)
rf.fit(X, y)
rf_importance = pd.Series(rf.feature_importances_, index=X.columns).sort_values(ascending=False)
print("\n Embedded Method (Random Forest Importance):\n", rf_importance.head(5))
```

```python
#3. Wrapper Method (Search-Based).Uses model performance to evaluate subsets of features (Forward/Backward/RFE).

from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import RFE

# Logistic Regression + RFE (Recursive Feature Elimination)
log_reg = LogisticRegression(max_iter=500)
rfe = RFE(log_reg, n_features_to_select=5)
rfe.fit(X, y)

selected_rfe = list(X.columns[rfe.support_])
print("\nWrapper Method (RFE with Logistic Regression) Selected Features:\n", selected_rfe)
```

```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor
from statsmodels.tools.tools import add_constant

# Step 1: Define features manually
features = [
    'Cost_of_the_Product', 'Customer_care_calls', 'Product_importance',
    'Prior_purchases', 'Weight_in_gms', 'Customer_rating',
    'Mode_of_Shipment_Ship', 'Warehouse_block_F', 'Gender',
    'Warehouse_block_C', 'Warehouse_block_D', 'Warehouse_block_B',
    'Mode_of_Shipment_Road', 'Discount_offered'
]

# Step 2: Subset your DataFrame
X = df[features].copy()

# Step 3: One-hot encode categorical variables
X = pd.get_dummies(X, drop_first=True)

# Step 4: Coerce all data to numeric (convert strings to NaN, etc.)
X = X.apply(pd.to_numeric, errors='coerce')

# Step 5: Drop rows with any NaNs or infinite values
X = X.replace([np.inf, -np.inf], np.nan)
X = X.dropna()

# Step 6: Add a constant column for intercept
X_const = add_constant(X)

# Step 7: Calculate VIF
vif_data = pd.DataFrame()
vif_data['Feature'] = X_const.columns
vif_data['VIF'] = [
    variance_inflation_factor(X_const.values.astype(np.float64), i)
    for i in range(X_const.shape[1])
]

# Step 8: Remove the constant (intercept) term from the result
vif_data = vif_data[vif_data["Feature"] != "const"]

# Step 9: Display result
print(vif_data.sort_values(by="VIF", ascending=False))
```

```python
# Step 1: Import Libraries
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.feature_selection import chi2
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv('C:/Users/DELL/Downloads/archive/Train.csv')
# Step 2: Load the Iris Dataset
iris = load_iris(as_frame=True)
df = iris.frame

# Optional: Check target distribution
print("Target value counts:\n", df['target'].value_counts())

# Step 3: Define Features (X) and Target (y)
X = df[['sepal length (cm)', 'sepal width (cm)', 
        'petal length (cm)', 'petal width (cm)']]
y = df['target']

# Step 4: Normalize the features to [0, 1] for Chi-squared test
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# Step 5: Apply Chi-squared test
chi2_vals, p_vals = chi2(X_scaled, y)

# Step 6: Create a DataFrame with results
chi2_results = pd.DataFrame({
    'Feature': X.columns,
    'Chi2 Value': chi2_vals,
    'p-value': p_vals
})

# Step 7: Add Decision Column (based on alpha = 0.05)
chi2_results['Reject Null Hypothesis (p <= 0.05)'] = chi2_results['p-value'] <= 0.05

# Step 8: Display Results
print("\nChi-Squared Test Results:")
print(chi2_results.sort_values(by="Chi2 Value", ascending=False))
```

```python
#1.Forward Selection using SequentialFeatureSelector (from sklearn)
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.model_selection import train_test_split
import pandas as pd

# Load data
iris = load_iris(as_frame=True)
X = iris.data
y = iris.target

# Model
model = LogisticRegression(max_iter=1000)

# Forward Selection
forward_selector = SequentialFeatureSelector(model, 
                                             direction='forward',
                                             n_features_to_select='auto', 
                                             scoring='accuracy', 
                                             cv=5)
forward_selector.fit(X, y)

# Results
selected_forward = X.columns[forward_selector.get_support()]
print("Forward Selection Chosen Features:")
print(selected_forward)
```

```python
#2.Backward Elimination using SequentialFeatureSelector
# Backward Elimination
backward_selector = SequentialFeatureSelector(model, 
                                              direction='backward',
                                              n_features_to_select='auto', 
                                              scoring='accuracy', 
                                              cv=5)
backward_selector.fit(X, y)

# Results
selected_backward = X.columns[backward_selector.get_support()]
print("\nBackward Elimination Chosen Features:")
print(selected_backward)
```

```python
#3.Recursive Feature Elimination (RFE)
from sklearn.feature_selection import RFE

# Recursive Feature Elimination
rfe_selector = RFE(estimator=model, n_features_to_select=2)
rfe_selector.fit(X, y)

# Results
selected_rfe = X.columns[rfe_selector.get_support()]
print("\nRecursive Feature Elimination (RFE) Chosen Features:")
print(selected_rfe)
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, classification_report

# Load Iris dataset
iris = load_iris(as_frame=True)
df_iris = iris.frame

X_iris = df_iris.drop('target', axis=1)
y_iris = df_iris['target']

# Split into train and test
X_train_iris, X_test_iris, y_train_iris, y_test_iris = train_test_split(X_iris, y_iris, test_size=0.3, random_state=42, stratify=y_iris)

# Train classifier
model_iris = RandomForestClassifier(random_state=42)
model_iris.fit(X_train_iris, y_train_iris)

# Predict
y_pred_iris = model_iris.predict(X_test_iris)

# Metrics
print("=== Iris Dataset Metrics ===")
print("Confusion Matrix:")
print(confusion_matrix(y_test_iris, y_pred_iris))
print(f"Accuracy: {accuracy_score(y_test_iris, y_pred_iris):.4f}")
print(f"Precision (macro): {precision_score(y_test_iris, y_pred_iris, average='macro'):.4f}")
print(f"Recall (macro): {recall_score(y_test_iris, y_pred_iris, average='macro'):.4f}")
print(f"F1 Score (macro): {f1_score(y_test_iris, y_pred_iris, average='macro'):.4f}")
print("\nClassification Report:")
print(classification_report(y_test_iris, y_pred_iris))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

df = pd.read_csv('C:/Users/DELL/Downloads/archive/Train.csv')
X = df.drop('Reached.on.Time_Y.N', axis=1)
X = pd.get_dummies(X, drop_first=True)
y = df['Reached.on.Time_Y.N']

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Define model
log_reg = LogisticRegression(max_iter=1000, random_state=42)

# Define parameter grid
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],   # inverse of regularization strength
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']  # solver compatible with l1 penalty
}

# Grid Search
grid_search = GridSearchCV(estimator=log_reg, param_grid=param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

print("Best params:", grid_search.best_params_)
print("Best CV accuracy:", grid_search.best_score_)

# Evaluate best model on test set
best_log_reg = grid_search.best_estimator_
y_pred = best_log_reg.predict(X_test)
print("Test accuracy:", accuracy_score(y_test, y_pred))
```
