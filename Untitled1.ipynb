pandas query

```python
#Customers with High Product Cost (> 200)
df_customer_analytics.query("Cost_of_the_Product > 200")
```

```python
#Female Customers Who Received More Than 30% Discount
df_customer_analytics.query("Gender == 'F' and Discount_offered > 30")
```

```python
#Customers Who Called Customer Care 4 or More Times
df_customer_analytics.query("Customer_care_calls >= 4")
```

```python
#Customers Who Received Their Order on Time
df_customer_analytics.query(" `Reached.on.Time_Y.N` == 1 ")
```

```python
df_customer_analytics.query("Gender == 'M' and Customer_rating >= 4 and Product_importance == 'low'")
```

```python
#Orders Shipped by Flight with Cost Between 150 and 200
df_customer_analytics.query("Mode_of_Shipment == 'Flight' and Cost_of_the_Product >= 150 and Cost_of_the_Product <= 200")
```

```python
#Customers with More Than 2 Prior Purchases and Rating Below 3
df_customer_analytics.query("Prior_purchases > 2 and Customer_rating < 3")
```

```python
#Customers with Light Shipments (Weight < 1500g) and Discount > 40
df_customer_analytics.query("Weight_in_gms < 1500 and Discount_offered > 40")
```

```python
#Shipments via Ship and rating is exactly 3
df_customer_analytics.query("Mode_of_Shipment == 'Ship' and Customer_rating == 3")
```

```python
#Customers who made more than 2 prior purchases and got low product importance
df_customer_analytics.query("Prior_purchases > 2 and Product_importance == 'low'")
```

PLOTS

```python
import matplotlib.pyplot as plt
import seaborn as sns
#matplotlib python library is for creating static, animated and interactive plots
#seaborn a statistical data visualization libary
```

```python
plt.figure(figsize=(7,5))#plot with a size of 7 inches wide and 5 inches tall.
sns.histplot(df_customer_analytics['Cost_of_the_Product'],bins=50,kde=True)
#bins split the data into given intervals on x-axis
#kde adds a smooth curve over the histogram to show the probability distribution of the data
```

```python
sns.countplot(x='Mode_of_Shipment',data=df_customer_analytics)
#a bar chart showing the count of each unique value in the Mode_of_Shipment
#countplot creates a bar plot of value counts for a categorical column
```

```python
sns.boxplot(x='Cost_of_the_Product', data=df_customer_analytics)
#creates a boxplot that shows the distribution of the Cost_of_the_Product column in the df_customer_analytics DataFrame.
```

```python
sns.violinplot(x='Cost_of_the_Product',data=df_customer_analytics)
#violinplot combines the features of both boxplot and kde plot
#plot that visualizes the distribution of the Cost_of_the_Product column
```

```python
sns.lineplot(x='Prior_purchases',y='Cost_of_the_Product',data=df_customer_analytics)
#lineplot() to draw a line chart that shows the relationship between:
#Prior_purchases (x-axis): Number of purchases made before
#Cost_of_the_Product (y-axis): Price of the product
```

```python
sns.scatterplot(x='Weight_in_gms',y='Cost_of_the_Product',data=df_customer_analytics,hue='Reached.on.Time_Y.N')
#creates a scatter plot, Blue dots = Delivered on time (1), Orange dots = Late (0)
```
